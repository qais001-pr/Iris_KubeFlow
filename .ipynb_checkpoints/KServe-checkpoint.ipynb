```python
from kfp import dsl
from kfp import compiler


# ============================================================
# COMPONENT 1
# LOAD IRIS DATA
# ============================================================

@dsl.component(
    base_image="python:3.11",
    packages_to_install=[
        "scikit-learn",
        "pandas"
    ]
)
def load_iris_data(
    dataset_output: dsl.Output[dsl.Dataset]
):

    import pandas as pd
    from sklearn.datasets import load_iris

    print("=" * 70)
    print("COMPONENT 1 — LOAD IRIS DATA")
    print("=" * 70)

    iris = load_iris()

    df = pd.DataFrame(
        iris.data,
        columns=[
            "sepal_length",
            "sepal_width",
            "petal_length",
            "petal_width"
        ]
    )

    df["target"] = iris.target

    print("Dataset shape:", df.shape)
    print("\nDataset:")
    print(df.head())

    df.to_csv(
        dataset_output.path,
        index=False
    )

    print(
        "\nDataset saved:",
        dataset_output.path
    )

    print("\nCOMPONENT 1 COMPLETED")


# ============================================================
# COMPONENT 2
# PREPROCESS
# ============================================================

@dsl.component(
    base_image="python:3.11",
    packages_to_install=[
        "pandas",
        "scikit-learn"
    ]
)
def preprocess_data(
    input_dataset: dsl.Input[dsl.Dataset],

    train_output: dsl.Output[dsl.Dataset],

    test_output: dsl.Output[dsl.Dataset]
):

    import pandas as pd

    from sklearn.model_selection import train_test_split

    print("=" * 70)
    print("COMPONENT 2 — PREPROCESS")
    print("=" * 70)

    df = pd.read_csv(
        input_dataset.path
    )

    print(
        "Input shape:",
        df.shape
    )

    train_df, test_df = train_test_split(

        df,

        test_size=0.20,

        random_state=42,

        stratify=df["target"]
    )

    print(
        "Train shape:",
        train_df.shape
    )

    print(
        "Test shape:",
        test_df.shape
    )

    train_df.to_csv(
        train_output.path,
        index=False
    )

    test_df.to_csv(
        test_output.path,
        index=False
    )

    print("\nTrain dataset saved")

    print("Test dataset saved")

    print("\nCOMPONENT 2 COMPLETED")


# ============================================================
# COMPONENT 3
# TENSORFLOW TRAINING
# ============================================================

@dsl.component(
    base_image="tensorflow/tensorflow:2.16.1",
    packages_to_install=[
        "pandas",
        "numpy"
    ]
)
def train_tensorflow_model(

    train_dataset: dsl.Input[dsl.Dataset],

    model_output: dsl.Output[dsl.Model],

    learning_rate: float = 0.001,

    hidden_units: int = 16,

    batch_size: int = 16,

    epochs: int = 30
):

    import os

    import numpy as np

    import pandas as pd

    import tensorflow as tf

    print("=" * 70)
    print("COMPONENT 3 — TENSORFLOW TRAINING")
    print("=" * 70)

    print(
        "TensorFlow version:",
        tf.__version__
    )

    print(
        "Learning rate:",
        learning_rate
    )

    print(
        "Hidden units:",
        hidden_units
    )

    print(
        "Batch size:",
        batch_size
    )

    print(
        "Epochs:",
        epochs
    )

    # --------------------------------------------------------
    # LOAD DATA
    # --------------------------------------------------------

    df = pd.read_csv(
        train_dataset.path
    )

    features = [
        "sepal_length",
        "sepal_width",
        "petal_length",
        "petal_width"
    ]

    X = df[
        features
    ].values.astype(
        np.float32
    )

    y = df[
        "target"
    ].values.astype(
        np.int32
    )

    print(
        "X shape:",
        X.shape
    )

    print(
        "Y shape:",
        y.shape
    )

    # --------------------------------------------------------
    # NORMALIZATION
    # --------------------------------------------------------

    normalizer = tf.keras.layers.Normalization()

    normalizer.adapt(X)

    # --------------------------------------------------------
    # MODEL
    # --------------------------------------------------------

    model = tf.keras.Sequential([

        tf.keras.layers.Input(
            shape=(4,),
            name="features"
        ),

        normalizer,

        tf.keras.layers.Dense(
            hidden_units,
            activation="relu"
        ),

        tf.keras.layers.Dense(
            8,
            activation="relu"
        ),

        tf.keras.layers.Dense(
            3,
            activation="softmax"
        )
    ])

    # --------------------------------------------------------
    # COMPILE
    # --------------------------------------------------------

    model.compile(

        optimizer=tf.keras.optimizers.Adam(
            learning_rate=learning_rate
        ),

        loss="sparse_categorical_crossentropy",

        metrics=[
            "accuracy"
        ]
    )

    print("\nModel created")

    # --------------------------------------------------------
    # TRAIN
    # --------------------------------------------------------

    history = model.fit(

        X,

        y,

        epochs=epochs,

        batch_size=batch_size,

        verbose=1
    )

    # --------------------------------------------------------
    # METRICS
    # --------------------------------------------------------

    accuracy = float(
        history.history["accuracy"][-1]
    )

    loss = float(
        history.history["loss"][-1]
    )

    print(
        "\nFinal training accuracy:",
        accuracy
    )

    print(
        "Final training loss:",
        loss
    )

    # --------------------------------------------------------
    # METADATA
    # --------------------------------------------------------

    model_output.metadata["framework"] = "tensorflow"

    model_output.metadata["dataset"] = "iris"

    model_output.metadata["accuracy"] = accuracy

    model_output.metadata["loss"] = loss

    model_output.metadata["learning_rate"] = learning_rate

    model_output.metadata["hidden_units"] = hidden_units

    model_output.metadata["batch_size"] = batch_size

    # --------------------------------------------------------
    # SAVEDMODEL
    # --------------------------------------------------------

    print("\nExporting SavedModel...")

    model_dir = model_output.path

    os.makedirs(
        model_dir,
        exist_ok=True
    )

    model.export(
        model_dir
    )

    # --------------------------------------------------------
    # VALIDATE SAVEDMODEL
    # --------------------------------------------------------

    saved_model_file = os.path.join(
        model_dir,
        "saved_model.pb"
    )

    variables_dir = os.path.join(
        model_dir,
        "variables"
    )

    print(
        "saved_model.pb exists:",
        os.path.exists(
            saved_model_file
        )
    )

    print(
        "variables exists:",
        os.path.exists(
            variables_dir
        )
    )

    if not os.path.exists(
        saved_model_file
    ):

        raise RuntimeError(
            "saved_model.pb was not created"
        )

    if not os.path.exists(
        variables_dir
    ):

        raise RuntimeError(
            "variables directory was not created"
        )

    print("\nSavedModel is KServe compatible")

    print("\nCOMPONENT 3 COMPLETED")


# ============================================================
# COMPONENT 4
# EVALUATE
# ============================================================

@dsl.component(
    base_image="tensorflow/tensorflow:2.16.1",
    packages_to_install=[
        "pandas",
        "numpy"
    ]
)
def evaluate_model(

    test_dataset: dsl.Input[dsl.Dataset],

    model_input: dsl.Input[dsl.Model],

    metrics_output: dsl.Output[dsl.Metrics]
):

    import numpy as np

    import pandas as pd

    import tensorflow as tf

    print("=" * 70)
    print("COMPONENT 4 — MODEL EVALUATION")
    print("=" * 70)

    # --------------------------------------------------------
    # LOAD TEST DATA
    # --------------------------------------------------------

    df = pd.read_csv(
        test_dataset.path
    )

    features = [
        "sepal_length",
        "sepal_width",
        "petal_length",
        "petal_width"
    ]

    X_test = df[
        features
    ].values.astype(
        np.float32
    )

    y_test = df[
        "target"
    ].values.astype(
        np.int32
    )

    print(
        "Test shape:",
        X_test.shape
    )

    # --------------------------------------------------------
    # LOAD SAVEDMODEL
    # --------------------------------------------------------

    print("\nLoading SavedModel...")

    model = tf.saved_model.load(
        model_input.path
    )

    print(
        "SavedModel loaded"
    )

    # --------------------------------------------------------
    # PREDICTION
    # --------------------------------------------------------

    infer = model.signatures[
        "serving_default"
    ]

    predictions = infer(
        tf.constant(X_test)
    )

    output_tensor = list(
        predictions.values()
    )[0]

    predicted_classes = tf.argmax(
        output_tensor,
        axis=1
    ).numpy()

    accuracy = float(
        np.mean(
            predicted_classes == y_test
        )
    )

    print(
        "\nTest accuracy:",
        accuracy
    )

    metrics_output.log_metric(
        "test_accuracy",
        accuracy
    )

    print("\nCOMPONENT 4 COMPLETED")


# ============================================================
# COMPONENT 5
# MINIO UPLOAD
# ============================================================

@dsl.component(
    base_image="python:3.11",
    packages_to_install=[
        "boto3"
    ]
)
def upload_model_to_minio(

    model_input: dsl.Input[dsl.Model],

    minio_endpoint: str,

    minio_access_key: str,

    minio_secret_key: str,

    bucket_name: str = "models",

    model_prefix: str = "iris/1"
):

    import os

    import boto3

    print("=" * 70)
    print("COMPONENT 5 — UPLOAD MODEL TO MINIO")
    print("=" * 70)

    print(
        "MinIO endpoint:",
        minio_endpoint
    )

    print(
        "Bucket:",
        bucket_name
    )

    print(
        "Model prefix:",
        model_prefix
    )

    # --------------------------------------------------------
    # MINIO CLIENT
    # --------------------------------------------------------

    s3 = boto3.client(

        "s3",

        endpoint_url=minio_endpoint,

        aws_access_key_id=minio_access_key,

        aws_secret_access_key=minio_secret_key,

        region_name="us-east-1"
    )

    # --------------------------------------------------------
    # CHECK BUCKET
    # --------------------------------------------------------

    try:

        s3.head_bucket(
            Bucket=bucket_name
        )

        print(
            "Bucket exists"
        )

    except Exception:

        print(
            "Creating bucket..."
        )

        s3.create_bucket(
            Bucket=bucket_name
        )

    # --------------------------------------------------------
    # UPLOAD
    # --------------------------------------------------------

    model_dir = model_input.path

    uploaded_files = 0

    for root, dirs, files in os.walk(
        model_dir
    ):

        for file in files:

            local_file = os.path.join(
                root,
                file
            )

            relative_file = os.path.relpath(

                local_file,

                model_dir
            )

            relative_file = relative_file.replace(
                "\\",
                "/"
            )

            object_name = (
                model_prefix
                + "/"
                + relative_file
            )

            print(
                "Uploading:",
                object_name
            )

            s3.upload_file(

                local_file,

                bucket_name,

                object_name
            )

            uploaded_files += 1

    # --------------------------------------------------------
    # VERIFY
    # --------------------------------------------------------

    print(
        "\nUploaded files:",
        uploaded_files
    )

    response = s3.list_objects_v2(

        Bucket=bucket_name,

        Prefix=model_prefix
    )

    print("\nMinIO objects:")

    if "Contents" in response:

        for obj in response["Contents"]:

            print(
                obj["Key"]
            )

    else:

        raise RuntimeError(
            "No model objects found in MinIO"
        )

    print("\nMODEL AVAILABLE AT:")

    print(
        f"s3://{bucket_name}/{model_prefix}"
    )

    print("\nCOMPONENT 5 COMPLETED")


# ============================================================
# COMPONENT 6
# KSERVE MODEL VALIDATION
# ============================================================

@dsl.component(
    base_image="python:3.11"
)
def validate_kserve_model(

    model_input: dsl.Input[dsl.Model]
):

    import os

    print("=" * 70)
    print("COMPONENT 6 — KSERVE MODEL VALIDATION")
    print("=" * 70)

    model_dir = model_input.path

    print(
        "Model directory:",
        model_dir
    )

    # --------------------------------------------------------
    # REQUIRED FILE
    # --------------------------------------------------------

    saved_model = os.path.join(

        model_dir,

        "saved_model.pb"
    )

    variables = os.path.join(

        model_dir,

        "variables"
    )

    print(
        "\nsaved_model.pb:",
        saved_model
    )

    print(
        "Exists:",
        os.path.exists(
            saved_model
        )
    )

    print(
        "\nvariables:",
        variables
    )

    print(
        "Exists:",
        os.path.exists(
            variables
        )
    )

    if not os.path.exists(
        saved_model
    ):

        raise RuntimeError(
            "KServe validation failed: "
            "saved_model.pb missing"
        )

    if not os.path.exists(
        variables
    ):

        raise RuntimeError(
            "KServe validation failed: "
            "variables directory missing"
        )

    print(
        "\nKServe validation SUCCESS"
    )

    print(
        "\nModel is ready for:"
    )

    print(
        "MinIO"
    )

    print(
        "KServe"
    )

    print(
        "TensorFlow Serving"
    )

    print("\nCOMPONENT 6 COMPLETED")


# ============================================================
# PIPELINE
# ============================================================

@dsl.pipeline(
    name="iris-kserve-validation",
    description="Iris TensorFlow model validation and MinIO upload pipeline"
)
def iris_kserve_pipeline(

    learning_rate: float = 0.001,

    hidden_units: int = 16,

    batch_size: int = 16,

    epochs: int = 30,

    minio_endpoint: str = "http://minio-service.kubeflow.svc.cluster.local:9000",

    minio_access_key: str = "minio",

    minio_secret_key: str = "minio123",

    bucket_name: str = "models",

    model_prefix: str = "iris/1"
):

    # --------------------------------------------------------
    # 1. LOAD
    # --------------------------------------------------------

    load_task = load_iris_data()

    # --------------------------------------------------------
    # 2. PREPROCESS
    # --------------------------------------------------------

    preprocess_task = preprocess_data(

        input_dataset=
        load_task.outputs[
            "dataset_output"
        ]
    )

    # --------------------------------------------------------
    # 3. TRAIN
    # --------------------------------------------------------

    train_task = train_tensorflow_model(

        train_dataset=
        preprocess_task.outputs[
            "train_output"
        ],

        learning_rate=learning_rate,

        hidden_units=hidden_units,

        batch_size=batch_size,

        epochs=epochs
    )

    # --------------------------------------------------------
    # 4. EVALUATE
    # --------------------------------------------------------

    evaluate_model(

        test_dataset=
        preprocess_task.outputs[
            "test_output"
        ],

        model_input=
        train_task.outputs[
            "model_output"
        ]
    )

    # --------------------------------------------------------
    # 5. MINIO
    # --------------------------------------------------------

    upload_model_to_minio(

        model_input=
        train_task.outputs[
            "model_output"
        ],

        minio_endpoint=minio_endpoint,

        minio_access_key=minio_access_key,

        minio_secret_key=minio_secret_key,

        bucket_name=bucket_name,

        model_prefix=model_prefix
    )

    # --------------------------------------------------------
    # 6. KSERVE VALIDATION
    # --------------------------------------------------------

    validate_kserve_model(

        model_input=
        train_task.outputs[
            "model_output"
        ]
    )


# ============================================================
# COMPILE
# ============================================================

compiler.Compiler().compile(

    pipeline_func=iris_kserve_pipeline,

    package_path="iris-kserve-validation.yaml"
)

print("=" * 70)

print(
    "PIPELINE COMPILED SUCCESSFULLY"
)

print(
    "File: iris-kserve-validation.yaml"
)

print("=" * 70)
```
